```python
import jVMC_exp.operator as op
import jVMC_exp
import jax
import jax.numpy as jnp
```

```python
L = 4
H = 0
for i in range(L):
    H += op.discrete.SigmaZ(i) * op.discrete.SigmaZ((i + 1) % L)
    H += op.discrete.SigmaX(i)

H.batch_size = 2 ** L // 5
```

```python
net = jVMC_exp.nets.RBM(L)
psi = jVMC_exp.vqs.NQS(net, L, 2**L)
sampler = jVMC_exp.sampler.ExactSampler(psi)
samples = sampler.samples
```
